### Train Disruptions
Als je deze dataset gebruikt is het belangrijk om te beseffen dat deze data gaat over verstoringen die door NS gecommuniceerd zijn. Niet elke trein die vertraagd of opgeheven is wordt door NS gecommuniceerd als een verstoring; de vuistregel die NS hanteert is dat een verstoring gecommuniceerd wordt wanneer meerdere treinen vertraagd of opgeheven zijn (dus een grote impact op de treindienst).

Het is ook belangrijk om te beseffen dat er sinds 2017 meer verstoringen gecommuniceerd worden, omdat NS toen een nieuw systeem heeft geïntroduceerd waarmee ze verstoringen eerder kunnen aankondigen (wat resulteert in meer verstoringen met een korte duur). Het vergelijken van het aantal verstoringen van 2017 met het aantal verstoringen in de jaren daarvoor is daardoor niet mogelijk (tenzij je rekening houdt met de toename van korte verstoringen).

De bron voor de verstoringen is altijd NS; de afdeling reisinformatie v
an NS monitort 24 uur per dag de treindienst om te zien of er verstoringen zijn. De storingsberichten in de open data zijn dezelfde berichten als op de borden op het station, de stationsomroep en op de Rijden de Treinen website en app.

https://www.rijdendetreinen.nl/open-data/treinstoringen

Alle treinritten in Nederland worden sinds 2019 opgeslagen in het treinarchief van Rijden de Treinen. Deze dataset bevat alle vertragingen, opgeheven treinen en dienstregelingswijzigingen van alle treinen in Nederland.

Deze dataset bevat alle reizigerstreinen in Nederland sinds 2019. De data wordt aangeboden als CSV-bestanden gecomprimeerd met Gzip.

Iedere rij in deze bestanden representeert een stop op een station. Iedere rit vertrekt vanaf en komt aan op een station (dus twee rijen). Voor iedere stop vind je de naam van het station, de aankomst- en vertrektijd, vertragingen en opgeheven ritten. De exacte betekenis van iedere kolom wordt hieronder uitgelegd.

De bron voor deze data is de real-time data van NS met live vertrektijden, actuele aankomsttijden en dienstregelingswijzigingen. Deze data wordt ook gebruikt in de app en website van Rijden de Treinen.

https://www.rijdendetreinen.nl/open-data/treinarchief

## NS Open Data URL's

```python
# ✅ Dataset base URLs (now specific per type)
base_url = "https://opendata.rijdendetreinen.nl/public"
page_urls = {
    "disruptions": f"{base_url}/disruptions/",
    "services": f"{base_url}/services/",
    "stations": f"{base_url}/stations/",
    "tariff-distances": f"{base_url}/tariff-distances/"
}

# Output folders
base_dir = target_root / "downloads/data/raw"
folders = {
    "disruptions": base_dir / "NS-disruptions",
    "services": base_dir / "NS-services",
    "stations": base_dir / "NS-stations",
    "tariff-distances": base_dir / "NS-tariff-distances"
}
for folder in folders.values():
    folder.mkdir(parents=True, exist_ok=True)
```

## Scrape download links

```python
FILE_RE = re.compile(r".*/?(disruptions|services)-(\d{4})\.(csv|csv\.gz)$", flags=re.IGNORECASE)
STATIONS_RE = re.compile(r".*/?stations-2023-09-nl\.csv$", flags=re.IGNORECASE)
TARIFF_RE = re.compile(r".*/?tariff-distances-2022-01(?:-ns-ns-international)?\.csv$", flags=re.IGNORECASE)

def get_dataset_links(page_urls):
    links = {key: [] for key in folders.keys()}

    for key, url in page_urls.items():
        try:
            print(f"Fetching links from: {url}")
            r = requests.get(url, timeout=20)
            r.raise_for_status()
            soup = BeautifulSoup(r.text, "html.parser")
            anchors = soup.find_all("a")

            for a in anchors:
                href = a.get("href")
                if not href:
                    continue
                href_abs = requests.compat.urljoin(url, href)

                # Match by dataset type
                if key in ["disruptions", "services"]:
                    m = FILE_RE.search(href_abs)
                    if m:
                        dataset_type, year = m.group(1).lower(), int(m.group(2))
                        if dataset_type == key and year >= 2019:
                            links[key].append(href_abs)

                elif key == "stations" and STATIONS_RE.search(href_abs):
                    links["stations"].append(href_abs)

                elif key == "tariff-distances" and TARIFF_RE.search(href_abs):
                    links["tariff-distances"].append(href_abs)

        except Exception as e:
            print(f"Error fetching {url}: {e}")

    # Sort and deduplicate
    for key in links:
        links[key] = sorted(set(links[key]))

    return links

# Fetch all datasets
links = get_dataset_links(page_urls)

# Summary
for key, urls in links.items():
    print(f"{key}: {len(urls)} files")
    for u in urls:
        print("   ", u)
```

```python
def download_file(url, base_dir):
    filename = url.split("/")[-1].lower()

    # Detect folder by keyword
    if "disruptions" in filename:
        out_path = base_dir / "NS-disruptions" / filename
    elif "services" in filename:
        out_path = base_dir / "NS-services" / filename
    elif "stations" in filename:
        out_path = base_dir / "NS-stations" / filename
    elif "tariff-distances" in filename:
        out_path = base_dir / "NS-tariff-distances" / filename
    else:
        out_path = base_dir / filename  # fallback

    if out_path.exists():
        return url, "exists", out_path

    try:
        with requests.get(url, stream=True, timeout=30) as r:
            r.raise_for_status()
            with open(out_path, "wb") as f:
                for chunk in r.iter_content(8192):
                    f.write(chunk)
        return url, "ok", out_path
    except Exception as e:
        return url, f"error: {e}", out_path
```

## Download all NS files (2min, download is est ~2GB, 20GB when unzipped)
NS files contains services, disruptions, stations, station distances

```python
# Flatten all links
all_links = sum(links.values(), [])
print(f"Total files to download: {len(all_links)}")

results = []
with ThreadPoolExecutor(max_workers=6) as ex:
    futs = [ex.submit(download_file, url, base_dir) for url in all_links]
    if tqdm:
        for f in tqdm(as_completed(futs), total=len(futs)):
            results.append(f.result())
    else:
        for f in as_completed(futs):
            results.append(f.result())

# Show summary
for url, status, path in results:
    print(f"{status:10} -> {path}")
```

## Merge all disruptions dataset into one file

```python
import duckdb
from pathlib import Path

# 1. Setup path
data_dir = Path("downloads/data/raw/NS-disruptions")
merged_disruptions_file = data_dir / "disruptions_merged_all.parquet"

# 2. Filter: Get ONLY files matching "disruptions-{year}.csv"
# We look for "disruptions-*.csv" and ensure the suffix is a number
csv_files = []
for f in data_dir.glob("disruptions-*.csv"):
    # Extract the part after "disruptions-"
    # e.g., "disruptions-2023" -> suffix_part is "2023"
    suffix_part = f.stem.replace("disruptions-", "")
    
    if suffix_part.isdigit():
        csv_files.append(str(f))

print(f"Found {len(csv_files)} yearly disruption CSV files to merge.")

if not csv_files:
    print("❌ No matching 'disruptions-{year}.csv' files found. Aborting.")
else:
    # 3. SQL Query
    # Convert python list to string representation: "['path/a.csv', 'path/b.csv']"
    files_sql_list = str(csv_files)

    # Note: We use read_csv with auto_detect=True to handle headers and types automatically
    query = f"""
    SELECT * FROM read_csv(
        {files_sql_list}, 
        union_by_name=True,
        filename=True,
        auto_detect=True,
        header=True
    )
    """

    # 4. Execute
    print("Merging CSVs into a single Parquet file...")

    try:
        duckdb.sql(f"COPY ({query}) TO '{merged_disruptions_file}' (FORMAT PARQUET);")
        print(f"✅ Done! Merged file saved to: {merged_disruptions_file}")
    except Exception as e:
        print(f"❌ Error merging files: {e}")
```

### Stations connections list
Validating consistency of NS API and RdT Stations list

```python
from collections import defaultdict

def build_filtered_adjacency_list(geojson_data, valid_codes):
    payload = geojson_data.get("payload", {})
    features = payload.get("features", [])
    
    adjacency_list = defaultdict(set)
    
    valid_set = set(valid_codes)

    for feature in features:
        props = feature.get("properties", {})
        
        raw_a = props.get("from")
        raw_b = props.get("to")
        
        # 1. Safety check: Ensure both exist before processing
        if raw_a and raw_b:
            station_a = raw_a.upper()
            station_b = raw_b.upper()

            # 2. Check A AND Check B independently
            if (station_a in valid_set) and (station_b in valid_set):
                
                adjacency_list[station_a].add(station_b)
                adjacency_list[station_b].add(station_a)

    return {k: sorted(list(v)) for k, v in adjacency_list.items()}

adj_list = build_filtered_adjacency_list(data, station_df_codes)

print(f"Number of stations in DataFrame: {len(station_df_codes)}")
print(f"Number of stations in Graph: {len(adj_list)}")
```

Api code to retrieve the weather data is an adaptation of the code available at https://pypi.org/project/openmeteo-requests/ and https://open-meteo.com/en/docs/historical-weather-api

```python
import openmeteo_requests
import requests_cache
import pandas as pd
from retry_requests import retry
import os
import time

# --- CONFIGURATION ---
# Create a folder for the data
OUTPUT_DIR = Path("downloads/data/raw/weather")
os.makedirs(OUTPUT_DIR, exist_ok=True)

# 2. Load and Prepare Station Data
folder_path = Path("downloads/data/raw/")
stations_csv_path = folder_path / "NS-stations/stations-2023-09-nl.csv"

# Read CSV, accounting for New Amsterdam station code issue, could just do this parameter instead of creating new parquet
stations_df = pd.read_csv(stations_csv_path, keep_default_na=False)

# Setting up API variables
weather_variables = [
    "temperature_2m", "rain", "snowfall", "snow_depth",
    "wind_speed_10m", "wind_gusts_10m", "soil_temperature_0_to_7cm"
]

timezone = "Europe/Amsterdam"

# --- API SETUP ---
cache_session = requests_cache.CachedSession('.cache', expire_after = -1)
retry_session = retry(cache_session, retries = 5, backoff_factor = 0.2)
openmeteo = openmeteo_requests.Client(session = retry_session)

# --- API ENDPOINT ---
# Utilize Customer Archive API endpoint if you have an API key
url = "https://customer-archive-api.open-meteo.com/v1/archive"
# Note: If you don't have an API key, you can use the free Open-Meteo endpoint
# url = "https://archive-api.open-meteo.com/v1/archive"

# --- MAIN LOOP ---
print(f"Starting download for {len(stations_df)} station locations...")

for idx, station in stations_df.iterrows():
    loc_id = station['id']
    file_path = f"{OUTPUT_DIR}/{loc_id}.parquet"
    
    # # 1. CHECKPOINTING: If file exists, skip!
    # if os.path.exists(file_path):
    #     print(f"Skipping {loc_id} (already exists)")
    #     continue

    print(f"Processing {loc_id}, Station: {station['name_long']}")

    # 2. PREPARE REQUEST
    params = {
        "latitude": station['geo_lat'],
        "longitude": station['geo_lng'],
        "start_date": "2019-01-01",
        "end_date": "2025-03-01",
        "hourly": weather_variables,
        "timezone": timezone,
        "apikey": api_key
    }

    try:
        # 3. CALL API
        # Note: We request one location at a time to keep memory footprint low
        responses = openmeteo.weather_api(url, params=params)
        response = responses[0] # We only requested one location

        # 4. PROCESS DATA
        hourly = response.Hourly()

        # Generate the time sequence ONCE into a variable
        time_range = pd.date_range(
            start = pd.to_datetime(hourly.Time(), unit = "s", utc = True),
            end = pd.to_datetime(hourly.TimeEnd(), unit = "s", utc = True),
            freq = pd.Timedelta(seconds = hourly.Interval()),
            inclusive = "left"
        )

        # Initialize dictionary with this time range
        # We call it 'date' here to be the index later
        hourly_data = {"date": time_range}

        # Iterate dynamically through requested variables
        # The Open-Meteo library returns variables in the order requested
        for i, var_name in enumerate(weather_variables):
            hourly_data[var_name] = hourly.Variables(i).ValuesAsNumpy()

        # Create DataFrame
        df = pd.DataFrame(data = hourly_data)
        
        # Add metadata (optional, but useful)
        df['rdt_id'] = loc_id
        df["time"] = time_range
        df["station_code"] = station["code"]
        df["station_name"] = station["name_long"]
        df["latitude"] = station["geo_lat"]
        df["longitude"] = station["geo_lng"]

        # 5. SAVE TO DISK IMMEDIATELY
        # Parquet is efficient for storage and retrieval, so we convert pandas to parquet
        df.to_parquet(file_path, compression='snappy')
        
        # 6. CLEAR MEMORY
        del df
        del hourly_data
        del station
        
    except Exception as e:
        print(f"ERROR processing {loc_id}: {e}")

print("All downloads complete.")
```

## Merging all train stations weather data (~400mb)

```python
import duckdb
from pathlib import Path

# 1. Setup path
weather_dir = Path("downloads/data/raw/weather")
merged_file = weather_dir / "weather_merged_all.parquet"

# 2. Filter: Get ONLY numeric parquet files (1.parquet, 125.parquet)
# We convert them to string paths immediately for DuckDB
numeric_files = [
    str(f) for f in weather_dir.glob("*.parquet") 
    if f.stem.isdigit()
]

print(f"Found {len(numeric_files)} numeric weather files to merge.")

if not numeric_files:
    print("❌ No numeric files found. Aborting.")
else:
    # 3. SQL Query
    # We pass the python list `numeric_files` directly into read_parquet.
    # DuckDB's Python client handles the list conversion automatically 
    # if you use f-strings carefully or bindings.
    
    # We construct a SQL-compatible list string: ['path/1.parquet', 'path/2.parquet']
    files_sql_list = str(numeric_files)

    query = f"""
    SELECT * FROM read_parquet(
        {files_sql_list}, 
        union_by_name=True,
        filename=True
    )
    """

    # 4. Execute
    print("Merging into a single Parquet file...")

    try:
        duckdb.sql(f"COPY ({query}) TO '{merged_file}' (FORMAT PARQUET);")
        print(f"✅ Done! Merged file saved to: {merged_file}")
    except Exception as e:
        print(f"❌ Error merging files: {e}")
```
